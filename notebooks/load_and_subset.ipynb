```python
#Drop irrelevant columns from mdata.obs
mdata.obs = (mdata.obs.drop(columns=['nFeature_RNA', 'nCount_SCT', 'nFeature_SCT', 
                                     'integrated_snn_res.0.5', 
                                     'integrated_snn_res.2', 'integrated_snn_res.0.1', 
                                     'integrated_snn_res.1', 'seurat_clusters',
                                     'celltype1', 'celltype2', 'celltype3', 'level2',
                                     'nCount_RNA', 'percent.mt', 'n_genes']))
                                    
#Replace all periods with underscores, make all column names lowercase
mdata.obs.columns = (mdata.obs.columns
                     .str.replace('.', '_', regex=False)
                     .str.lower())
```

```python
#Drop irrelevant columns from mdata.obs
#Values will be recalculated in scanpy
#Features is the same as the index
mdata.var = (mdata.var.drop(columns=['highly_variable', 'dispersions',
                                     'features', 'dispersions_norm',
                                     'n_cells', 'means']))
```

```python
#Drop all uns data and umap coordinates
del mdata.uns
del mdata.obsm["X_umap"]
```
