```python
import scanpy as sc
import anndata as ad
import pandas as pd
import numpy as np
from pathlib import Path
import seaborn as sns
import harmonypy as hm
import scanorama
import matplotlib.pyplot as plt

script_loc = Path('01_load_and_preprocess.py').resolve().parent
data_loc = script_loc.parent / "data"
resources_loc = script_loc.parent / 'resources'
outputs_loc = script_loc.parent / 'outputs_full'
```

```python
hcdata = sc.read_h5ad(data_loc / 'intermediate/hc_integrated.h5ad')
hpdata = sc.read_h5ad(data_loc / 'intermediate/hp_integrated.h5ad')

scdata = sc.read_h5ad(data_loc / 'intermediate/sc_integrated.h5ad')
spdata = sc.read_h5ad(data_loc / 'intermediate/sp_integrated.h5ad')
```

```python
#Calculate the sex lisi values for hcdata
mtadata = hcdata.obs[['sex']].copy()
harmony_sexlisi = (hm.compute_lisi(X=hcdata.obsm['X_pca_harmony'],
                                metadata=mtadata,
                                label_colnames=['supercluster_name']))
df_sexlisi = pd.DataFrame()
df_sexlisi['harmony'] = pd.DataFrame(harmony_sexlisi)

#Calculate the clisi values for scdata
mtadata = scdata.obs[['sex']].copy()
scanorama_sexlisi = (hm.compute_lisi(X=scdata.obsm['X_scanorama'],
                                metadata=mtadata,
                                label_colnames=['supercluster_name']))
df_sexlisi['scanorama'] = pd.DataFrame(scanorama_sexlisi)

#Save clisi scores to csv
df_sexlisi.to_csv(outputs_loc / 'tables/c_sexlisi_scores.csv', index=False)
```

```python
#Create a bar plot of sex lisi scores across methods
df_sexlisi = df_sexlisi.melt(var_name="method", value_name="lisi_score")
(sns.barplot(data=df_sexlisi, x='method', y='lisi_score', errorbar='sd')
    .set(title='Caudate Integration Sex LISI Scores', xlabel='Integration Method', ylabel='Sex LISI Score'))
plt.savefig(outputs_loc / 'exploration_figures/c_sexlisi_scores.png', dpi=300)
plt.show()
```
